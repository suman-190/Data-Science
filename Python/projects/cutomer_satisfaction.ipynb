```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 100  # number of samples

data = {
    'Customer_ID': range(1, n+1),
    'Product_Quality': np.random.randint(1, 6, size=n),
    'Service_Quality': np.random.randint(1, 6, size=n),
    'Price_Fairness': np.random.randint(1, 6, size=n)
}

df = pd.DataFrame(data)
df['Satisfaction_Score'] = df[['Product_Quality', 'Service_Quality', 'Price_Fairness']].mean(axis=1).round(2)
df['Recommend_Friend'] = np.where(df['Satisfaction_Score'] > 3.5, 'Yes', 'No')
df['Repurchase_Intention'] = np.where(df['Satisfaction_Score'] > 3.0, 'Yes', 'No')
df['Loyalty_Score'] = (df['Satisfaction_Score'] + np.where(df['Repurchase_Intention'] == 'Yes', 1, 0)).round(1)

print(df.head())
```
