```python
print("Hello")
# formatted string literal
# The f in print(f"...") stands for "formatted string literal" or f-string, which was introduced in Python 3.6.

# It allows you to embed expressions inside string literals using curly braces {}. These expressions are evaluated at runtime and then formatted using the result.

number = int(input("Enter a number"))

i=1
while i<=10:
    print(f"{number} * {i}  = {number*i}")
    i=i+1

```
