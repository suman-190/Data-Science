```python
# what are functions
# function is block of code which take some input and  produce some output
# abstraction

def sum(a,b):
    print(a+b)

sum(2,4)
```

```python
# Give number is even or odd
def is_even(i):
    '''This Function will check the number is true or not'''
    if(i%2==0):
        return "Even"
    else:
        return "Odd"

print(is_even(5))
# Accessign documentation of function
print(is_even.__doc__)

# Calling a function inside loop

for i in range(5):
    print(i,is_even(i))
```

### Type of Argunment
- Default Argument

- Positional Argument

- Key Word Argument   

```python
# Default argument
# Positional argument

def power(a,b=2):
    return a**b

print(power(2))
# keyword argument
print(power(b=4,a=2))
```

```python
# allows us to pass the varible number of arguments

def sum_number(*args):
    print(type(args))
    sum = 0
    for i in args:
        sum=sum+i
    return sum    

sum_number(2,3,4,5,6,7,8)
```

```python
# allows us to pass the varible number of keyword arguments
def user_info(**kwargs):
    print(kwargs)
    for key,val in kwargs.items():
        print(key,val)

    for key in kwargs.keys():
        print(key)

    for val in kwargs.values():
        print(val)
        
            

         
    


user_info(name="Suman Bhandari",age=23)    
```

```python
def multiple_argu(val,*args,**kwargs):
    print(val)
    print(args)
    print(kwargs)

multiple_argu(23,34,564,34,name="Suman",age="23")
```
