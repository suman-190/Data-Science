```python
# A Tuple in python is similer to a list. The diffrence between the two is that we cannot chnage the elements of a tuple once it is assigned whereas we can change the elements of a list

# Tuple is a class =>  Group of Variable and functions
# tuple is iterable
# Tuple is Immutable
# Tuple elements are indexed
# Tuple is Sequence
# Tuple can store Hetrogeneous elements





t1 = (2,2,2,3,4,6,7)
print(t1.count(2))

print(t1[-1])
print(t1)
t3 = (10,)
print(type(t3))
print(t3)

t4 = 10,30,40
print(type(t4))


```

```python
r1  = range(10,20)
# print(r1[3])


# Accessing Tuple ELements
t1 = (26,2,2,2,3,4,6,7)

# for i in t1:
#     print(i,end=" ") 


# Built in Methods
print(len(t1))
print(min(t1))
print(max(t1))
print(sorted(t1,reverse=True)) 
# List of Sorted elements
print(sum(t1))


```

```python
t1 = (10,20)
t2 = (11,22,33)
print(t1+t2)
print(t1*3)


# Comparison Operator
print(t1 > t2)
print(t1 == t2)


# Tuple Object Methods
print(t1.index(20))
print(t1.count(10))
```
